

#Test condition

np.where(condition,value_if_true,value_if_false)



```python
b = np.where(a>2,a,-1)
print(b)
```

```python
temps = np.array([28,32,35,30,25,27,22,33,36,29,39,26,24])
label = np.where(temps > 30 , "Hot" , "Cold")
print(label)
```

```python
marks = np.array([35,50,67,78,88,45])
result = np.where(marks > 60 , "Pass", "Fail")
print(result)
```

```python
marks = np.array([[45,33,67],[78,89,81],[39,50,48],[61,73,63]])
result = np.where (marks > 50 ,"Pass" ,"Fail")
print(result)
```

#Fancy Indexing - > To print multiple index at one time

```python
a = np.array([10,20,30,40,50,60,70])
print(a[5])
b = a[[1,3,5]]
print(b)
```

#np.argwhere(): Get indices of **True** values



```python
even = np.argwhere(a%3 == 0)
print(even)
even.shape
```

#Flatten -> convert 2D to 1D

```python
c = even.flatten()
print(c)
c.shape
```
