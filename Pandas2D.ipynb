# Filtering

````python
data = {"Name":["Mia","JC","CK","IC"],
        "Age":[25,19,45,30],
        "Salary":[4500,3000,5000,4700]}
df = pd.DataFrame(data)
print(df)
````

````python
df[ df["Age"] > 27]
````

````python
df[(df["Age"] >= 25) & (df["Age"] <= 30) ]
df[df["Salary"] != 3000 ]
````

# Grouping using .groupby()

````python
data = {"Name":["Mia","JC","CK","IC","Evonne","Caster","Tresure","Joko","Moko"],
        "City":["RGN","MDY","TGI","TGI","RGN","RGN","MDY","MON","DWE"],
        "Age":[25,19,45,29,30,18,23,34,27],
        "Salary":[4500,3000,5000,4700,5000,2300,3200,5300,4000]}
df = pd.DataFrame(data)
print(df)
````

````python
df.groupby("City")["Salary"].mean()
````

````python
df.groupby("City")["Name"].count()
````

````python
df.groupby("City")["Age"].max()
````

````python
df.groupby("City")["Age"].min()
````

# Handle missing values

```
using .fillna() or .dropna()
```



````python
data = {"Name":["Aung","Chaw","Su","Hla","Moe"],
        "Age":[25,np.nan,28,np.nan,35],
        "Salary":[3000,4500,np.nan,5000,np.nan],
        "City": ["RGN","DWE",None,"MDY","RGN"]}

df = pd.DataFrame(data)
print(df)
````

````python
df["Age"].fillna(df["Age"].mean())
````

````python
df.dropna(subset=["Salary"], inplace = True)
print(df)
````
