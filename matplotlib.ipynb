# Line Chart - suitable for Time series data

```python
x = ["Jan","Feb","Mar","Apr", "May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"]
y = [10, 25,15, 45, 60,55, 70,70,75,85,99,80]
plt.plot(x,y,"o-")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly sales")
plt.show()

plt.scatter(x,y,color = "green")
plt.show()
```

```python
dx = [1,2,3,4,5]
dy = [2,4,6,4,2]
dy1 = [4,2,4,1,5]
dy2 = [3,2,1,6,7]
plt.plot(dx,dy,"o-")
plt.plot(dx,dy1,"*-")
plt.plot(dx,dy2)
plt.show()
```

```python
x = np.linspace(0,100,10)
y = np.sin(x)
plt.plot(x,y,label= 'sin(x)',color = 'red', linestyle = '--')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.legend()
#plt.grid()
plt.show()
```

```python
dx = [1,2,3,4,5]
dy = [2,4,6,4,2]
dy1 = [4,2,4,1,5]
dy2 = [3,2,1,6,7]
plt.scatter(dx,dy)
plt.scatter(dx,dy1)
plt.scatter(dx,dy2)
plt.show()
```

```python
months = np.arange(1,13)
Yangon = np.random.randint(60,180,size= 12)
Mandalay = np.random.randint(50,150,size = 12)
NayPyiTaw = np.random.randint(55,155, size = 12)
rainfall_data = np.array([Yangon,Mandalay,NayPyiTaw])
print(rainfall_data)
monthly_avg = rainfall_data.mean(axis = 0)
print(monthly_avg)
plt.plot(months,monthly_avg)
plt.xlabel("Months")
plt.ylabel("Average Rainfall(mm)")
plt.title("Average monthly Rainfall")
plt.show()
```

# Box Plot - to identify whether there is an outlier

```python
java =[78,85,92,70,88]
database = [80,82,75,95]
web = [72,68,80,77,85]
AI = [65,70,75,60,72]
data = [java, database, web,AI]
plt.boxplot(data)
plt.title("Student scores distribution by subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.show()
```

```python
score = [10,45,50,52,55,58,60,61,62,65,67,68,70,72,74,76,80,85,110,120] # showing 2 distinct outliers
plt.boxplot(score)
plt.title("BoxPlot for test score")
plt.ylabel("Scores")
plt.show()
```

```python
data = [np.random.normal(0,std,100) for std in range(1,4)] # 1-3 -> (0,1,100),(0,2,100),(0,3,100)
#print(data)
plt.boxplot(data,vert=True, patch_artist = True)
plt.xticks ([1,2,3],["std=1","std=2","std=3"])
#plt.grid()
plt.title("BoxPlot for 3 distribution")
plt.show()
```

```python
classA = [45,50,52,55,58,60,61,62,63,65,70,120]
classB = [60,62,64,66,67,68,70,72,74,75]
classC = [30,30,40,45,50,55,88]
data= [classA,classB,classC]
plt.boxplot(data,patch_artist=True)
plt.xticks([1,2,3],["Class A","Class B","Clacc C"])
plt.title("Comparison of Test Scores")
plt.ylabel("Scores")
plt.grid()
plt.show()
```

```python
lables = ["Python","Java","C++","Javascript"]
counts = [40,25,20,15]
plt.pie(counts,labels=lables,autopct="%1.1f%%", startangle = 90) #autopct = 1.1f means one 0 after (.) / startangle = 90 means the angle will start from 90' angle
plt.axis("equal")
plt.show()
```

```python
x = np.linspace(0,2*np.pi ,100) # 1 2
                                # 3 4
#print(x)
fig,axs = plt.subplots(2,2,figsize = (10,6)) # (2,2) means 2 rows 2 columns
axs [0,0].plot(x,np.sin(x)); axs[0,0].set_title("sin(x)")
axs [0,1].plot(x,np.cos(x)); axs[0,1].set_title("cos(x)")
axs[1,0].plot(x,np.sin(2*x)); axs[1,0].set_title("sin(2x)")
axs[1,1].plot(x,np.exp(-x)); axs[1,1].set_title("exp(-x)")
fig.tight_layout()
plt.show()
```
