
# TWO major data structure
1. Series - one dimensional labeled array , list
2. Data frame - 2 dimensional ( rows and columns)



```python
import numpy as np
import pandas as pd
data = [10,20,30,40,50]
s = pd.Series(data)
print(s)
```

```python
data = [10,20,30,40,50]
#s = pd.Series(data, index=["A","B","C","D","E"]) # giving the index name rather than auto index numbering
s = pd.Series(data, name ="Data") # giving title
print(s)
```

```python
#DICTIONARY -> key, value / key-> index , value -> series data
data = {"A":10,"B":20,"C":30}
print(data)
s = pd.Series(data)
print(s)
```

```python
marks={"Kelvin":85,"Helen":92,"Mia":78}
s = pd.Series(marks)
print(s)
```

```python
data = {"A":10,"B":20,"C":30}
s = pd.Series(data, index = ["C","A","B"]) # cutomized print style
print(s)
```

```python
employee=("John Smith",29,55000,True)
print(employee)

s = pd.Series(employee, index = ["Name","Age","Salary","Full Time"])
print(s)
```

```python
arr = np.array([10,20,30,40,50])
print(arr)
s = pd.Series(arr)
print(s)
```

```python
arr = np.array(["Apple",1200,1.5,True], dtype = object)
s = pd.Series(arr, index = ["Product","Price","Weight","Instock"])
print(s)
```

```python
temp = np.array([28.5,29.1,30.2,23.0])
s = pd.Series(temp, index = ["9AM","10AM","11AM","12PM"])
print(s)
```

# Retrieving Data from Pandas series

```python
s = pd.Series([10,20,30,40,50])
print(s[0])
print(s[[2,4]])
```

```python
s = pd.Series([100,200,300,400], index = ["A","B","C","D"])
print(s["B"])
print(s[["A","D"]]) # print location of b and c
print(s[1:3])
```

```python

#using .loc[] -> same like index based
print(s.loc["B"])
print(s.loc["A":"C"])#print location of A until C

#index location
print(s.iloc[1])

```

# Filtering

```python
s = pd.Series([10,20,30,40,50])
filtered = s[s>30]
print(filtered)
```

```python
filtered = s[(s > 20 & (s <= 50))]
print(filtered)
```

```python
s = pd.Series(["Apple","Melon","Mango"])
filtered = s[s.isin(["Apple","Melon"])]
print(filtered)

filtered1 = s[s.isin(["Grape","Banana"])]
print(filtered1)
```

```python
#Vectorized operation

data1 = [1,2,3]
s1 = pd.Series(data1)

data2 = [4,5,6]
s2 = pd.Series(data2)

print(s1 + s2)
```

```python
prices = pd.Series([1500,200,500,1200] , index = ["Milk","Coffee","Maltmailk","Bread"])
print(prices)

```

```python
new_prices = prices * 1.10
print(new_prices)

expensive= prices[prices >= 1500]
print(expensive)
```
